# Sales Analysis Q3 2025

This notebook analyzes quarterly sales performance across regions.

## Key Metrics
- Total Revenue
- Units Sold
- Customer Acquisition Cost

```python
import pandas as pd

sales_data = {
    'region': ['North', 'South', 'East', 'West'],
    'revenue': [125000, 98000, 142000, 87000],
    'units': [340, 280, 410, 220]
}
df = pd.DataFrame(sales_data)
print(df.describe())
```

## Observations

| Region | Revenue Rank | Notes |
|--------|-------------|-------|
| East   | 1st         | Highest growth YoY |
| North  | 2nd         | Stable performer |
| South  | 3rd         | Needs attention |
| West   | 4th         | New market expansion |

```python
# Calculate CAC per region
marketing_spend = {'North': 15000, 'South': 12000, 'East': 18000, 'West': 9000}
df['cac'] = df['region'].map(marketing_spend) / df['units']
df.sort_values('cac', ascending=True)
```
